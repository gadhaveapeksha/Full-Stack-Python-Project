```python
x1 = input('Enter the 1st number') #whenevery you works in input function it always give you string 
y1 = input('Enter the 2nd number') # it wont understand as arithmetic operator
z1 = x1 + y1
print(z1)
```

```python
type(x1)
type(y1)
```

```python
x1 = input('Enter the 1st number') #whenevery you works in input function it always give you string 
a1 = int(x1)
y1 = input('Enter the 2nd number') # it wont understand as arithmetic operator
b1 = int(y1)
z1 = a1 + b1
print(z1)
```
